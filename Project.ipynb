### EDA

### Merging Tables

```python
dfs = [life_expectancy, life_expectancy_for_males, life_expectancy_for_females, death_rate, infant_mortality, neonatal_mortality]
```

```python
df = reduce( lambda left, right: pd.merge(left, right, on=['country_isocode', 'country_name']), dfs)
```

```python
df.head()
```

### Questions

##### *Kindly note that all this data from 2022

### 1) Which country has minimum life expected average in Africa and which one has the maximum?

```python
print("Minimum life expected years Country:")
df[df['life_expected_years'] == df['life_expected_years'].min()][['country_name', 'life_expected_years']]
```

```python
print("Maximum life expected years Country:")
df[df['life_expected_years'] == df['life_expected_years'].max()][['country_name', 'life_expected_years']]
```

### 2) Is there any country with Male life expected more than Female life expected?

```python
df[df['male_life_expected_years'] > df['female_life_expected_years']][['country_name', 'life_expected_years']]
```

### 3) Which country has the highest death rate and which one has the lowest?

#### *Kindly note that these numbers are per 1000

```python
print("Highest death rate:")
df[df['death_rate'] == df['death_rate'].max()][['country_name', 'death_rate']]
```

```python
print("Lowest death rate:")
df[df['death_rate'] == df['death_rate'].min()][['country_name', 'death_rate']]
```

### 4) Create infant mortality/country graph

#### *Kindly note that these numbers are per 1000

```python
sorted_infant = df[['country_name', 'infant_mortality']].sort_values(by='infant_mortality', ascending= False)
sorted_infant.plot.bar(x='country_name', figsize = (20, 8))
```

### 5) Which countries are better to live in?

##### Here we will divide each column to 4 divisions and assign a score according to its value, the highest the score, the low the mortality

```python
df['score1'] = pd.qcut(df['life_expected_years'], 4, labels=[1, 2, 3, 4])
df['score2'] = pd.qcut(df['death_rate'], 4, labels=[4, 3, 2, 1])
df['score3'] = pd.qcut(df['infant_mortality'], 4, labels=[4, 3, 2, 1])
df['score4'] = pd.qcut(df['neonatal_mortality'], 4, labels=[4, 3, 2, 1])
df.head(10)
```

```python
df['score1'] = df['score1'].astype(int)
df['score2'] = df['score2'].astype(int)
df['score3'] = df['score3'].astype(int)
df['score4'] = df['score4'].astype(int)
```

```python
df.info()
```

```python
df['total_score'] = df['score1'] + df['score2'] + df['score3'] + df['score4']
df[['country_name', 'total_score']].sort_values(by= 'total_score', ascending= False)
```
